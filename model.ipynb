# Machine Learning

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

import joblib
```

```python
df.head()
```

```python
df.columns
```

```python
df_model=df.drop(['nameOrig','nameDest','isFlaggedFraud'], axis=1)
```

```python
df_model.head()
```

```python
catagorical = ["type"]
numarical = ["amount","oldbalanceOrg","newbalanceOrig","oldbalanceDest","newbalanceDest",]
```

```python
# Feature Engineering
y = df_model["isFraud"]
X = df_model.drop(["isFraud"], axis=1)
```

```python
#split Data
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, stratify=y)
```

```python
#ColumsTransformer
preprocessors = ColumnTransformer(
    transformers= [
        ("num", StandardScaler(), numarical),
        ("cat", OneHotEncoder(drop='first'), catagorical)
    ],
    remainder= 'drop'
)
```

```python
#preprocessor
```

```python
#pipeline
Pipeline = Pipeline([
    ("prep", preprocessors),
    ("clf", LogisticRegression(class_weight="balanced", max_iter=1000))

    ])
```

```python
#pipeline
```

```python
Pipeline.fit(X_train,
             y_train)
```

```python
y_pred = Pipeline.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
confusion_matrix(y_test, y_pred)
```

```python
Pipeline.score(X_test, y_pred)
```

#Save Model

```python
joblib.dump(Pipeline,"fraud_detection_Pipeline.pkl")
print("Model_Saved_successfully")
```
